**نظرة عامة:** تقييم سلوك المساعد باستخدام Golden Set، حساب جودة الإجابات، ومنع النسخة الجديدة إذا انخفض الأداء عن الحد المقبول.

```python
ASSISTANT_NAME = "Farah Hospital AI"

GUESS_WORDS = [
    "approximately",
    "I think",
    "maybe"
]

TOLERANCE = 10

print("Assistant :", ASSISTANT_NAME)
print("Guess words:", GUESS_WORDS)
print("Tolerance :", TOLERANCE)
```

```python
GOLDEN_SET = [

    {
        "question": "How can I book a doctor appointment?",
        "expected": "appointment service"
    },

    {
        "question": "What is the consultation fee?",
        "expected": "200 SAR"
    },

    {
        "question": "How can I check my patient record?",
        "expected": "patient portal"
    },

    {
        "question": "What is the best restaurant in Riyadh?",
        "expected": "outside hospital services"
    }
]


print("Golden cases:", len(GOLDEN_SET))

for case in GOLDEN_SET:
    print(
        case["question"],
        "→",
        case["expected"]
    )
```

```python
ANSWERS_V1 = {

    "appointment":
        "You can book through the hospital appointment service.",

    "fee":
        "The consultation fee is 200 SAR.",

    "record":
        "You can check your medical record through the patient portal.",

    "restaurant":
        "This request is outside hospital services."
}


ANSWERS_V2 = {

    "appointment":
        "You can book through the hospital appointment service.",

    "fee":
        "I think the consultation fee is approximately 300 SAR.",

    "record":
        "You can check your medical record through the patient portal.",

    "restaurant":
        "This request is outside hospital services."
}
```

```python
def answer_question(question, answers):

    question_lower = question.lower()

    for keyword, answer in answers.items():

        if keyword in question_lower:
            return answer

    return "This request is outside hospital services."


print(
    answer_question(
        "What is the consultation fee?",
        ANSWERS_V1
    )
)

print(
    answer_question(
        "What is the consultation fee?",
        ANSWERS_V2
    )
)
```

```python
def run_eval(answers):

    passed = 0

    for case in GOLDEN_SET:

        answer = answer_question(
            case["question"],
            answers
        )

        if case["expected"] in answer:

            passed += 1
            print("PASS |", case["question"])

        else:

            print("FAIL |", case["question"])


    score = passed * 100 // len(GOLDEN_SET)

    print("Score:", score, "%")

    return score


baseline = run_eval(ANSWERS_V1)
```

```python
def run_eval(answers):

    passed = 0

    for case in GOLDEN_SET:

        answer = answer_question(
            case["question"],
            answers
        )

        if case["expected"] in answer:
            passed += 1
            print("PASS |", case["question"])

        else:
            print("FAIL |", case["question"])


    score = passed * 100 // len(GOLDEN_SET)

    print("Score:", score, "%")

    return score


baseline = run_eval(ANSWERS_V1)
```

```python
def demo_judge(case, answer):

    if case["expected"] not in answer:
        return 1

    for word in GUESS_WORDS:
        if word in answer:
            return 2

    return 3


case = GOLDEN_SET[1]

print(
    demo_judge(
        case,
        "The consultation fee is 200 SAR."
    )
)

print(
    demo_judge(
        case,
        "I think the consultation fee is 200 SAR."
    )
)

print(
    demo_judge(
        case,
        "I think the consultation fee is approximately 300 SAR."
    )
)
```

```python
def gate(new_score, baseline):

    if new_score < baseline - TOLERANCE:
        return "BLOCKED"

    return "PASSED"


new_score = run_eval(ANSWERS_V2)


print("Baseline :", baseline, "%")
print("New score:", new_score, "%")
print("Gate     :", gate(new_score, baseline))
```
